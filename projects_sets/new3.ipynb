```python
from langchain_huggingface.embeddings import HuggingFaceEndpointEmbeddings

embeddings = HuggingFaceEndpointEmbeddings(
    model="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
store = InMemoryStore(index={"embed":embeddings,'dims':384})

```

```python
namespace = ("users","u1")
```

```python
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User prefers detailed explanations with examples"})
store.put(namespace, "3", {"data": "User likes answers in bullet points"})
store.put(namespace, "4", {"data": "User prefers responses in simple language"})
store.put(namespace, "5", {"data": "User is interested in Python programming"})
store.put(namespace, "6", {"data": "User prefers real-world examples"})
store.put(namespace, "7", {"data": "User likes short summaries at the end"})
store.put(namespace, "8", {"data": "User prefers step-by-step explanations"})
store.put(namespace, "9", {"data": "User is learning about AI and machine learning"})
store.put(namespace, "10", {"data": "User prefers code examples in responses"})
store.put(namespace, "11", {"data": "User likes comparisons between concepts"})
store.put(namespace, "12", {"data": "User prefers Hindi explanations occasionally"})
store.put(namespace, "13", {"data": "User is a beginner in programming"})
store.put(namespace, "14", {"data": "User prefers practical use cases"})
store.put(namespace, "15", {"data": "User likes visual analogies for understanding"})
store.put(namespace, "16", {"data": "User prefers minimal technical jargon"})
```

```python
items = store.search(namespace,query="what is hte user currently learning",limit=1)
for item in items:
    print(item.value)
```
